```python
X1 = [0,1,0,1]
X2 = [0,0,1,1]

def test(x1, x2):
    if x1 == 1 and x2 == 1:
        return 1
    else:
        return 0

def disp_results(func):
    for i in range(4):
        Y = func(X1[i], X2[i])
        print(f"{X1[i]}, {X2[i]} = {Y}")

disp_results(test)
```

```python
def and_test(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.8
    ans = w1 * x1 + w2 * x2
    if ans > theta:
        return 1
    else:
        return 0

disp_results(and_test)
```
